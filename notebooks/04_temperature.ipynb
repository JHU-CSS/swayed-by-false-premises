# 04 · Robustness to sampling temperature

Reproduces the temperature table in the appendix: Claude Sonnet on the same 100 items per dataset at temperature 0 (greedy, one run) and at 0.25, 0.5, 0.75 and 1.0 (three runs each). Cells are mean ± standard deviation over runs.

```python
import sys
from pathlib import Path

BASE_DIR = Path("..").resolve()
sys.path.insert(0, str(BASE_DIR))

DATA_DIR = BASE_DIR / "data"
OUTPUT_DIR = DATA_DIR / "outputs"
FIG_DIR = BASE_DIR / "figures"
TABLE_DIR = BASE_DIR / "tables"
EXPECTED_DIR = BASE_DIR / "expected"

FIG_DIR.mkdir(exist_ok=True)
TABLE_DIR.mkdir(exist_ok=True)
```

```python
import pandas as pd

from sway import revision as rv
from sway.checks import compare
```

```python
rv.configure(OUTPUT_DIR, FIG_DIR, TABLE_DIR)
summ, overall = rv.temp_tables()

compare(summ, pd.read_csv(EXPECTED_DIR / "temperature_by_dataset.csv"), ["Dataset", "Temperature"], ["S_mean", "S_std", "n_runs", "n_questions"], label="Temperature by dataset")
compare(overall, pd.read_csv(EXPECTED_DIR / "temperature_overall.csv"), ["Temperature"], ["S_mean", "S_std", "n_runs"], label="Temperature overall")
```

```python
table = rv.temperature_table(summ, overall)
table.to_csv(TABLE_DIR / "table_temperature.csv")
table
```
